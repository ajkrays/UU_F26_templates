## The standard stack: importing packages + reading in the data

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px

df = pd.read_csv('ForeignGifts_edu.csv',low_memory=False)
df.head()
```

## Groupby

As we learned, `.groupby` is a super useful way to aggregate the data and explore summary statistics. We can assign the output to a new dataframe, just taking the n largest amounts. Resetting the index just changes the index to integers starting at 0. 

<div class="alert alert-block alert-warning">
<b>Question 1:</b>  Which country gives the most money, in total?
</div>


```python
# Grouping by gifter to avoid repeatedly computing
grouped_gifter = df.groupby("Country of Giftor")["Foreign Gift Amount"]
```

```python
# We group by country of giftor, select the amount column, and compute the sum for each grouped unit
# Finally, we sort by descending to display the largest first
grouped_gifter.sum().sort_values(ascending=False).head(10)
```

Qatar gives the most money, with $2706240869 in total donations.

<div class="alert alert-block alert-warning">
<b>Question 2:</b>  Which country initiates the most gifts, by count?
</div>


```python
# We again group by giftor and select foreign gift amount
# However, this time we perform the size aggregation to count total rows
# Finally, we sort descending to see the largest counts
grouped_gifter.size().sort_values(ascending=False)
```

England initiates the mosts gifts, with 3655 in total.

<div class="alert alert-block alert-warning">
<b>Question 3:</b>  Which country gives the largest gifts, on average?
</div>

```python
# We group by country of giftor, select the amount column, and compute the average for each grouped unit
# Finally, we sort by descending to display the largest first
grouped_gifter.mean().sort_values(ascending=False)
```

Bermuda gives the largest gifts, with 7.68 x 10^6 dollars on average.

### Cross-tabulation
First, use `pd.crosstab` to look at the relationship between outcoming gifts from countries and incoming gifts to institutions. 

We can crosstab on gift totals, like so:

```python
pd.crosstab(df["Country of Giftor"], df["Institution Name"])
```

This isn't very helpful, given that there are so many combinations. We can similarly do the same but with donation amounts:

```python
pd.crosstab(df["Country of Giftor"], df["Institution Name"], values=df["Foreign Gift Amount"], aggfunc="mean")
```

### Flow of information using alluvial plots

To better visualize which countries money is flowing from and which institutions it is flowing into, we can use an alluvial diagram. There is a built-in version in plotly that we'll use.

```python
import plotly.graph_objects as go

# giftor = 'Giftor Name'
# recipi = 'Institution Name'
# flor = 'Foreign Gift Amount'

giftor = 'Giftor Name'
recipi = 'Institution Name'
flow = 'Foreign Gift Amount'
N = 20

flows = (
    df.groupby([giftor, 
                recipi])
      [flow]
      .sum()
      .nlargest(N)
      .reset_index()
)

labels = (
    flows[giftor].tolist()
    + flows[recipi].tolist()
)

labels = list(dict.fromkeys(labels))

fig = go.Figure(
    go.Sankey(
        node=dict(label=labels),
        link=dict(
            source=flows[giftor]
                        .map(labels.index),
            target=flows[recipi]
                        .map(labels.index),
            value=flows[flow]
        )
    )
)

fig.show()

```

We can also do it by quantity of donations:

```python
import plotly.graph_objects as go

# giftor = 'Giftor Name'
# recipi = 'Institution Name'
# flor = 'Foreign Gift Amount'

giftor = 'Giftor Name'
recipi = 'Institution Name'
N = 20

flows = (
    df.groupby([giftor, 
                recipi])
      .size()
      .nlargest(N)
      .reset_index()
)

labels = (
    flows[giftor].tolist()
    + flows[recipi].tolist()
)

labels = list(dict.fromkeys(labels))

fig = go.Figure(
    go.Sankey(
        node=dict(label=labels),
        link=dict(
            source=flows[giftor]
                        .map(labels.index),
            target=flows[recipi]
                        .map(labels.index),
            value=flows[0]
        )
    )
)

fig.show()
```
