```python
import numpy as np
import pandas as pd
from scipy.stats import norm
```

# Question 4

```python
df = pd.read_csv("data/ames_prices.csv")
```

```python
df["Central Air"] = (df["Central.Air"] == "Y").astype(int)
```

```python
df["Central Air"].value_counts()
```

```python
p_hat = sum(df["Central.Air"] == "Y") / len(df)
p_hat
```

```python
len(df)
```

## Part A

Let's generate the boostrap distribution:

```python
rng = np.random.default_rng(seed=42)
```

```python
R = 2000
results = []

for _ in range(R):
    idx = rng.integers(0 , len(df), len(df))
    p_hat_star = df.loc[idx, "Central Air"].mean()
    results.append(p_hat_star)

results = np.array(results)
```

## Part B

Now, let's estimate the standard error:

```python
standard_error = np.std(results, ddof=1)
standard_error
```

## Part C

Now, let's construct the confidence intervals at $90\%, $95\%$, and $99\%$ using the centered-normal approach.

```python
# 90% CI
low = -1.645 * standard_error + p_hat
high = 1.645 * standard_error + p_hat

print(f"90% CI: [{low}, {high}]")
```

```python
# 95% CI
low = -1.96 * standard_error + p_hat
high = 1.96 * standard_error + p_hat

print(f"95% CI: [{low}, {high}]")
```

```python
# 99% CI
low = -2.576 * standard_error + p_hat
high = 2.576 * standard_error + p_hat

print(f"99% CI: [{low}, {high}]")
```

## Part D

Now, let's perform the two-sided hypothesis test by computing the p-value. Note that each corresponding threshold of the hypothesis test is equivalent to one of the CIs above (a C% CI is equivalent to an $\alpha = 1 - C$ two-sided test), so we should expect them to line up accordingly. 

```python
z = (p_hat - 0.9) / standard_error
p = 2 * norm.cdf(-abs(z))

print(z, p)
```

This agrees with our corresponding confidence intervals, where all three intervals ($0.90, 0.95, 0.99$) do not contain the hypothesized value $0.9$. A more thorough treatment is given in part f.

## Part F

In context, our estimate, $\hat{p}$, is 0.933, meaning that approximately $93.3\%$ of our sample Ames, Iowa houses contains central air. For confidence intervals, recall that a $C\%$ confidence interval may be interpreted as follows: if we were to repeat the same procedure (bootstrap to identify standard error and then construct a corresponding interval around our sample estimate using the associated critical value for our confidence) using many different random samples, $C\%$ of these intervals would contain the true population proportion. Since we only construct one of these intervals at three different confidences, we cannot be certain that the true population proportion falls within each of our intervals, though they still provide us with plausible values in which the true population proportion may take on. Thus, we have $90\%$ confidence that the true proportion of houses with central air is between $0.925$ and $0.941$, $95\%$ confidence that the true proportion of houses with central air is between $0.924$ and $0.942$, and $99\%$ confidence that the true proportion of houses with central air is between $0.921$ and $0.945$. Since none of these intervals contain our hypothesized proportion of $0.90$, the data suggests that our true population proportion is not equal to $0.90$. This is confirmed by the corresponding hypothesis tests; as mentioned above, a C% CI is equivalent to an $\alpha = 1 - C$ two-sided hypothesis test, meaning that the p-value falls below our critical threshold exactly when the hypothesized value is not contained in the associated confidence interval. For instance, since $0.90$ was not included in our $90\%$ confidence interval, we should expect that our p-value will fall at or below the $\alpha = 0.1$ threshold; the similar statement applies for the pairs ($95\%$, $0.05$) and ($99\%$, $0.01%$). As computed above, the standardized z-score is $7.21$, and the p-value is $0.0000000000005510798747816234$. The p-value represents the probability of obtaining a sample with a proportion atleast this extreme, assuming the null hypothesis is true. Since $0.0000000000005510798747816234 \leq 0.1$, we reject the null hypothesis at $\alpha = 0.1$; a similar statement applies for $\alpha = 0.05$ and $\alpha = 0.01$, as anticipated. Thus, the data suggests (but does not prove) that the true proportion of Ames, Iowa houses with central air is not equal to $0.90$. In particular, it seems that the true proportion may be higher, though it is poor form to conclude this given that we set-up a two-sided test; we are technically doubling our type 1 error because we did not declare this one-sided inequality before conducting the test.

For confidence intervals, recall that a $C\%$ confidence interval may be interpreted as follows: if we were to repeat the same procedure (bootstrap to identify standard error and then construct a corresponding interval around our sample estimate using the associated critical value for our confidence) using many different random samples, $C\%$ of these intervals would contain the true population correlation. Since we only construct one of these intervals at three different confidences, we cannot be certain that the true population correlation falls within each of our intervals, though they still provide us with plausible values in which the true population correlation may take on. Thus, we have $90\%$ confidence that the true correlation of arcsinh(area) and arcsinh(price) for houses is between $0.706$ and $0.740$, $95\%$ confidence that the true correlation of arcsinh(area) and arcsinh(price) for Ames, Iowa houses is between $0.703$ and $0.743$, and $99\%$ confidence that the true correlation of arcsinh(area) and arcsinh(price) for houses is between $0.697$ and $0.749$. Since none of these intervals contain our hypothesized correlation of $0$, the data suggests that our true population correlation for Ames, Iowa houses is not equal to $0$. As computed above, the standardized z-score is $69.848$, and the p-value is 0. The p-value represents the probability of obtaining a sample with a correlation atleast this extreme, assuming the null hypothesis is true. Since $0 \leq 0.1$, we reject the null hypothesis at $\alpha = 0.1$; a similar statement applies for $\alpha = 0.05$ and $\alpha = 0.01$, as anticipated by the equivalence with the corresponding $1 - \alpha$ confidence interval. Thus, the data suggests (but does not prove) that the true correlation between arcsinh(area) and arcsinh(price) of Ames, Iowa houses is not equal to $0$. In particular, it seems that the true correlation may be positive, though it is poor form to conclude this given that we set-up a two-sided test; we are technically doubling our type 1 error because we did not declare this one-sided inequality before conducting the test.

When we bootstrap, we are implicitly leveraging the fact that our sample (in this case, pairs of (arcsinh(area), arcsinh(height))) is an approximate stand-in for the population; this holds for a random sample from the true population. If we mix our pairs, we are destroying the integrity of our sample points, and we no longer have valid sample points from the population we are attempting to approximate. That is, instead of sampling from the emprical joint distribution (which is a stand-in for the true joint distribution whose correlation sampling distribution we are interested in), we are sampling from a joint empirical distribution that is the product of the marginal empirical distributions; these two distributions are equal if and only if the two variables are independent, which we do not know apriori. Thus, the distribution generated is mostly likely not approximating the intended sampling distribution. 

Still, this regime can have application in other contexts. Moreover, resampling separately assumes that the two components of interest are independent of one another. As a result, the bootstrap distribution generated under this regime is a null, where the two variables are indpendent, and we can see where our sample correlation from the original data sits on it to suggest whether they are independent or not.
